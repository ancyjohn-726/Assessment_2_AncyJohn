```python
# Q1. NumPy Array Creation

import numpy as np
array_1=np.array([10,20,30,40,50])
print("The array:",array_1)
print("Shape of the array:",array_1.shape)
print("Data type of the array:",array_1.dtype)
```

```python
# Q2. NumPy Array Operations

array_2=np.array([5,10,15,20,25])
added_array=array_2 + 5
multiplied_array=array_2 * 2
print("Original Array:",array_2)
print("After adding 5:",added_array)
print("After multiplying by 2:",multiplied_array)
```

```python
# Q3. NumPy Statistical Operations

array_3=np.array([12,15,18,20,25,30])
mean_value=np.mean(array_3)
median_value=np.median(array_3)
min_value=np.min(array_3)
max_value=np.max(array_3)
std_value=np.std(array_3)
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
print(f"Minimum: {min_value}")
print(f"Maximum: {max_value}")
print(f"Standard deviation: {std_value:.4f}")
```

```python
# Q4. NumPy Filtering

array_4=np.array([10,25,30,45,50,65,70])
filtered_array=array_4[array_4 > 40]
print("Value greater than 40:",filtered_array)
```

```python
# Q10. Handling Missing Values

df_missing=pd.DataFrame({'Name': ['A', 'B', 'C', 'D'],
                         'Age': [21, np.nan, 23, 22],
                         'Score': [80, 75, np.nan, 90]})
df_missing['Age']=df_missing['Age'].fillna(df_missing['Age'].mean())
df_missing['Score']=df_missing['Score'].fillna(df_missing['Score'].mean())
print("DataFrame After added missing value\n")
print(df_missing)
```

SECTION B

```python
# Q11. Data Loading and Exploration

df_students=pd.read_csv('/content/student_performance.csv')
print("First 5 Records :\n")
print(df_students.head(5))
print("\nLast 5 Records :\n")
print(df_students.tail(5))
print(f"\nNumber of rows and columns :\n\n{df_students.shape[0]} rows, {df_students.shape[1]} columns")
print("\nColumn Names :\n\n", df_students.columns.tolist())
print("\nData Types :\n")
print(df_students.dtypes)
print("\nStatistical Summary :\n")
print(df_students.describe())
```

```python
# Q12. Filtering and Sorting

final_score_75=df_students[df_students['Final_Score']>75]
print("Students with Final_Score greater than 75:\n")
print(final_score_75)
attendance_75=df_students[df_students['Attendance']<75]
print("\nStudents with Attendance less than 75:\n")
print(attendance_75)
data_science_students=df_students[(df_students['Department']=='Data Science') & (df_students['Final_Score']>80)]
print("\nData Science students with Final_Score greater than 80:\n")
print(data_science_students)
sorted_students=df_students.sort_values(by='Final_Score',ascending=False)
print("\nSorted Students by Final_Score:\n")
print(sorted_students)
```

```python
# Q13. Feature Creation

new_column=['Python_Score','SQL_Score','Pandas_Score','NumPy_Score']
df_students['Average_Technical_Score']=df_students[new_column].mean(axis=1)
def performance(score):
    if score >=90:
        return 'Excellent'
    elif score >=75:
        return 'Good'
    elif score >=60:
        return 'Average'
    else:
        return 'Needs Improvement'
df_students['Performance']=df_students['Average_Technical_Score'].apply(performance)
print(df_students[['Name','Average_Technical_Score','Performance']].head())

```

```python
# Q14. GroupBy Analysis

grouped_df=df_students.groupby('Department').agg({'Name':'count','Final_Score':['mean','max','min']})
grouped_df.columns=['Number_of_Students','Average_Final_Score','Maximum_Final_Score','Minimum_Final_Score']
sorted_df=grouped_df.sort_values(by='Average_Final_Score',ascending=False)
print(sorted_df)
```

```python
# Q15. NumPy and Pandas Integration

scores = df_students["Final_Score"].dropna().values
print("Mean:\n", scores.mean())
print("\nMedian:\n", np.median(scores))
print("\nStandard Deviation:\n", scores.std())
print("\n25th Percentile:\n", np.percentile(scores, 25))
print("\n75th Percentile:\n", np.percentile(scores, 75))
print("\nStudents with Final_Score above the overall mean:\n")
display(df_students[df_students["Final_Score"] > scores.mean()])
```

```python
# Q16. Student Performance Analysis
# Load Data

df=pd.read_csv('/content/student_performance.csv')

# A. Data Cleaning

df.fillna(df.mean(numeric_only=True),inplace=True)
```

```python
# B. Feature Engineering

subjects=['Python_Score','SQL_Score','Pandas_Score','NumPy_Score']
df['Average_Score']=df[subjects].mean(axis=1)
df['Pass_Status']=np.where(df['Average_Score'] >= 50,'Pass','Fail')
```

```python
# C. Analysis

print("Top 5 Students:\n", df.nlargest(5, 'Average_Score')[['Name', 'Average_Score']])
print("\nDepartment having the highest average score:", df.groupby('Department')['Average_Score'].mean().idxmax())
print("\nAttendance by Department:\n\n", df.groupby('Department')['Attendance'].mean())
print("\nNumber of students passed and failed:\n\n", df['Pass_Status'].value_counts())
```

```python
# D. Summary DataFrame

summary_df=df.groupby('Department').agg(Total_Students=('Student_ID','count'),
                                        Average_Score=('Average_Score','mean'),
                                        Average_Attendance=('Attendance','mean'),
                                        Pass_Count=('Pass_Status',lambda x: (x =='Pass').sum()),
                                        Fail_Count=('Pass_Status',lambda x: (x =='Fail').sum())).reset_index()
print("\nSummary:\n", summary_df)
```
